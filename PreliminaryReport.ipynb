#### Hypothesis: Comedy films will be more commonly found in the ranking list for Top Grossing Movies of 2025 and earn more money than horror films due to how they make people feel.

```python
import pandas as pd
import requests

# load in the link from the website
html = requests.get(
    "https://www.the-numbers.com/market/2025/top-grossing-movies",
    headers={"User-Agent": "Mozilla/5.0"} #I needed to use AI for some help with this part , as the code does not run without it - I was getting an error.
).text

#read the link
df = pd.read_html(html)[0]

# get rid of the columns I do not want - the number of tickets sold and the distributor for the movie.
df = df.drop(columns=["Tickets Sold"], errors="ignore")
df = df.drop(columns=["Distributor"], errors="ignore")

#show the table
display(df)

# used for a later part after the visuals below to show there are more higher grossing horror films in top 25 
df.head(25)
```

```python
# make the table for the frequency of the genres appearing
Table = df["Genre"].value_counts().reset_index()
Table.columns = ["Genre", "Number of Movies"]

#show the table
display(Table)
```

```python
# get the counts for the number of commedy movies and horror movies
comedy_df = df[df["Genre"].str.contains("Comedy", case=False, na=False)]
horror_df = df[df["Genre"].str.contains("Horror", case=False, na=False)]
comedy_count = len(comedy_df)
horror_count = len(horror_df)

#print the values and name them certain titles
print("Comedy Movies:", comedy_count)
print("Horror Movies:", horror_count)

# ** I used AI for some help with this, as when I ran my code above it would give me an error message 
#stating "Cannot mask with non-boolean array containing NA / NaN values", and I ended up needing to 
#add "case=False, na=False" into the first 2 lines to resolve this and get my comedy movie and horror movie counts.
```

```python
import matplotlib.pyplot as plt

# Frequency comparison
plt.bar(["Comedy", "Horror"], [comedy_count, horror_count], color=['pink', 'black'])
plt.title("Number of Comedy vs Horror Movies")
plt.ylabel("Movie Count")
plt.show()
```
